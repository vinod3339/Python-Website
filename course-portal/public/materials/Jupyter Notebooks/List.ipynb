```python
numbers = [1, 2, 3, 4, 5]
numbers.append(6)
print(numbers)
```

```python
numbers = [1, 2, 3, 4, 5]
numbers.insert(2, 10)
print(numbers)
```

```python
numbers = [1, 2, 3, 4, 5]
numbers.remove(2)
print(numbers)
```

```python
numbers = [1, 2, 3, 4, 5]
numbers.pop(3)
print(numbers)
```

```python
numbers = [1, 2, 3, 4, 5, 2, 2]
count = numbers.count(2)
print(count)
```

```python
numbers = [5, 2, 9, 1, 5, 6]
numbers.sort()
print(numbers)
```

```python
numbers = [1, 2, 3, 4, 5]
numbers.reverse()
print(numbers)
```
