```python
#ask A1.1: Write a program that:
# 1. Stores your age, the current year, and your birth year in three integer variables.
# 2. Prints the type of each variable using type().
# 3. Calculates and prints your age in the year 2050

age = 25
current_year = 2024
birth_year = current_year - age

print("Type of age variable:", type(age))
print("Type of current_year variable:", type(current_year))
print("Type of birth_year variable:", type(birth_year))

age_in_2050 = 2050 - birth_year
print("Your age in the year 2050 will be:", age_in_2050)
```

```python
# Task A2.1: Write a program that:
# 1. Stores your first name and last name in two separate string variables.
# 2. Concatenates them into a full name with a space in between.
# 3. Prints the full name in UPPERCASE, lowercase, and Title Case.
# 4. Prints the length of the full name.
# 5. Prints the first character and the last character of the full name

first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name

print("Full name in UPPERCASE:", full_name.upper())
print("Full name in lowercase:", full_name.lower())
print("Full name in Title Case:", full_name.title())
print("Length of the full name:", len(full_name))
print("First character of the full name:", full_name[0])
print("Last character of the full name:", full_name[-1])
```

```python
# Task A3.1: Write a program that:
# 1. Declares two boolean variables, is_raining and has_umbrella.
# 2. Prints their type using type().
# 3. Prints the result of is_raining and has_umbrella, is_raining or has_umbrella, and not is_raining.

is_raining = True
has_umbrella = False

print("Type of is_raining variable:", type(is_raining))
print("Type of has_umbrella variable:", type(has_umbrella))
print("is_raining and has_umbrella:", is_raining and has_umbrella)
print("is_raining or has_umbrella:", is_raining or has_umbrella)
print("not is_raining:", not is_raining)    
```

```python
# <!-- Task B1.1: Take two integers a = 23 and b = 6. 
# Write a program to compute and print the result of all
# seven arithmetic operators listed above on a and b -->

a = 23
b = 6
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)
```

```python
# Task B2.1: Take two integers m and n entered by the user (use input() and int()). Print the result of
# comparing them with all six comparison operators

m = int(input("Enter the first integer (m): "))
n = int(input("Enter the second integer (n): "))

print("m == n:", m == n)
print("m != n:", m != n)
print("m > n:", m > n)
print("m < n:", m < n)
print("m >= n:", m >= n)
print("m <= n:", m <= n)
```

```python
# Task B3.1: Start with a variable score = 50. Apply each assignment operator one after another (in the
# order listed above), printing the value of score after every step.
score = 50
print("Initial score:", score)
score += 10
print("After += 10:", score)
score -= 5
print("After -= 5:", score)
score *= 2
print("After *= 2:", score)
score /= 4
print("After /= 4:", score)
score //= 3
print("After //= 3:", score)
score %= 7
print("After %= 7:", score)
```

```python
# Task B4.1: A student is eligible for a scholarship if their percentage is above 75 AND attendance is
# above 90%. Write a program that takes percentage and attendance as input and prints whether the
# student is eligible, using logical operators.

percentage = float(input("Enter your percentage: "))
attendance = float(input("Enter your attendance percentage: "))

if percentage > 75 and attendance > 90:
    print("You are eligible for the scholarship.")
else:
    print("You are not eligible for the scholarship.")
```

```python
# Task B5.1: Take two integers p = 12 and q = 10. Print the result of p & q, p | q, p ^ q, ~p, p << 2, and p
# >> 2. Convert p and q to their binary form using bin() and verify the bitwise results manually.
p = 12
q = 10

print("p & q:", p & q)
print("p | q:", p | q)
print("p ^ q:", p ^ q)
print("~p:", ~p)
print("p << 2:", p << 2)
print("p >> 2:", p >> 2)

print("Binary form of p:", bin(p))
print("Binary form of q:", bin(q))
```

```python
# Task B6.1: Create a list of five fruits. Ask the user to enter a fruit name and check whether it is in the list
# or not, using the in and not in operators.

fruits = ["apple", "banana", "cherry", "date", "elderberry"]
user_fruit = input("Enter a fruit name: ")
if user_fruit in fruits:
    print(f"{user_fruit} is in the list of fruits.")
else:
    print(f"{user_fruit} is not in the list of fruits.")
```

```python
# Task B7.1: Create two lists with the same content, list1 and list2, and a third variable list3 = list1. Use
# the is and is not operators (along with id()) to check and explain which variables refer to the same
# object in memory.

list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print("list1 is list2:", list1 is list2)
print("list1 is list3:", list1 is list3)
print("ID of list1:", id(list1))
print("ID of list2:", id(list2))
print("ID of list3:", id(list3))
```

```python
# Challenge Task (Optional)

# Write a single Python program for a simple login validator: it stores a predefined username (string) and
# password (string), takes user input for both, uses comparison and logical operators to check the
# credentials, and uses a boolean variable login_success to store and print the final result.
username = "admin"
password = "password"

user_username = input("Enter your username: ")
user_password = input("Enter your password: ")

if user_username == username and user_password == password:
    login_success = True
else:
    login_success = False

print("Login successful:", login_success)
```
