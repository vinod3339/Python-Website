```python
#CHALLENGE — Write a function  is_valid_identifier(name)  that checks (without using  str.isidentifier() ) whether agiven string could be a valid Python identifier, based on the rules (starts with letter/underscore, contains only letters/digits/underscores, not a keyword).

def is_valid_identifier(name):
    # Check if the name is empty
    if not name:
        return False

    # Check if the first character is a letter or underscore
    if not (name[0].isalpha() or name[0] == '_'):
        return False

    # Check if all characters are letters, digits, or underscores
    for char in name:
        if not (char.isalnum() or char == '_'):
            return False

    # Check if the name is a reserved keyword
    import keyword
    if keyword.iskeyword(name):
        return False

    return True

name = input("Enter a string to check if it's a valid Python identifier: ")
if is_valid_identifier(name):
    print(f"'{name}' is a valid Python identifier.")
else:
    print(f"'{name}' is not a valid Python identifier.")
```

```python
#Write a program that takes multiple values in a single input line separated by spaces (e.g.,  10 20 30 ), splits them, converts them to integers, and prints their sum.

numbers = input("Enter multiple numbers separated by spaces: ")
# Split the input string into a list of strings
number_list = numbers.split()
# Convert the list of strings to a list of integers
int_list = [int(num) for num in number_list]
# Calculate the sum of the integers
total_sum = sum(int_list)
print("Sum of the numbers:", total_sum) 
```

```python
#CHALLENGE — Take a list of 3 subject marks as input on one line, calculate and print the average, formatted to 2 decimal places

marks_input = input("Enter marks for 3 subjects separated by spaces: ")
marks_list = [float(mark) for mark in marks_input.split()]
average = sum(marks_list) / len(marks_list)
print(f"Average marks: {average:.2f}")
```

```python
#Rewrite Task 2 using the  argparse  module instead of  sys.argv , with named arguments  --num1  and  --num2 , and a  --help  message

import argparse

parser = argparse.ArgumentParser(description="Add two numbers")
parser.add_argument("--num1", type=int, help="First number")
parser.add_argument("--num2", type=int, help="Second number")

args = parser.parse_args()

if args.num1 is None or args.num2 is None:
    parser.print_help()
else:
    print(f"The sum of {args.num1} and {args.num2} is {args.num1 + args.num2}")
```
