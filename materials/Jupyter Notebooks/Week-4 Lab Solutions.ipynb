```python
#Create a list of numbers and demonstrate the use of append(), insert(), extend(), remove(), pop(), sort(), reverse(), count(), and index() methods, printing the list after each operation.

nums = [5, 2, 8, 1]

print("Initial list:", nums)

nums.append(7)
print("After append():", nums)

nums.insert(2, 9)
print("After insert():", nums)

nums.extend([4, 6])
print("After extend():", nums)

nums.remove(1)
print("After remove():", nums)

popped_value = nums.pop()
print("After pop():", nums, "Popped value:", popped_value)

nums.sort()
print("After sort():", nums)

nums.reverse()
print("After reverse():", nums)

print("Count of 2:", nums.count(2))
print("Index of 2:", nums.index(2))
```

```python
#Write a program to remove duplicate elements from a list without using the set() function.

# Example list with duplicate values
lst = [1, 2, 2, 3, 4, 4, 5, 3]

unique_list = []

for value in lst:
    if value not in unique_list:
        unique_list.append(value)

print("Original list:", lst)
print("List without duplicates:", unique_list)
```

```python
#Write a program to find the maximum, minimum, and sum of elements in a list without using built-in max(), min(), or sum() functions.

maximum = numbers[0]
minimum = numbers[0]
total = 0

for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num
    total += num

print("Maximum:", maximum)
print("Minimum:", minimum)
print("Sum:", total)
```

```python
#Given a tuple containing a nested list, modify the nested list and explain (in a comment) why this is possible despite tuples being immutable.
# Create a tuple containing a nested list
nested_tuple = (["apple", "banana"], "grape")

# Modify the nested list in place
nested_tuple[0].append("cherry")

print(nested_tuple)

# Tuples are immutable because you cannot assign a new element or rebind the tuple itself,
# but the list inside the tuple is a mutable object. The tuple stores a reference to that list,
# so changing the list's contents is allowed even though the tuple cannot be replaced.
```

```python
#Write a program to merge two dictionaries into one using the update() method and the | operator(Python 3.9+).

dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

# Using update() method
dict1.update(dict2)
print("Merged dictionary using update():", dict1)

# Using | operator (Python 3.9+)
dict3 = {"a": 1, "b": 2}
dict4 = {"c": 3, "d": 4}
merged_dict = dict3 | dict4
print("Merged dictionary using | operator:", merged_dict)
```

```python
#Given a dictionary of item names and prices, write a program to find the item with the highest price and the item with the lowest price

items = {"apple": 1.2, "banana": 0.8, "orange": 1.5, "grape": 2.0}
highest_item = max(items, key=items.get)
lowest_item = min(items, key=items.get)
print(f"Item with highest price: {highest_item} at ${items[highest_item]}")
print(f"Item with lowest price: {lowest_item} at ${items[lowest_item]}")
```

```python
#Write a program that counts the frequency of each character in a string and stores the result in a dictionary

text = "hello world"
char_freq = {}
for char in text:
    if char in char_freq:
        char_freq[char] += 1
    else:
        char_freq[char] = 1
print("Character frequencies:", char_freq)
```

```python
#Given two sets of numbers, write a program to find their union, intersection, difference, and symmetric difference.

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

union_set = set1.union(set2)
intersection_set = set1.intersection(set2)
difference_set = set1.difference(set2)
symmetric_difference_set = set1.symmetric_difference(set2)

print("Union:", union_set)
print("Intersection:", intersection_set)
print("Difference:", difference_set)
print("Symmetric Difference:", symmetric_difference_set)
```

```python
#Write a program to check whether one set is a subset or superset of another set using issubset() and issuperset().

set1 = {1, 2, 3}
set2 = {1, 2, 3, 4, 5}

is_subset = set1.issubset(set2)
is_superset = set2.issuperset(set1)

print("Is set1 a subset of set2?", is_subset)
print("Is set2 a superset of set1?", is_superset)
```

```python
#Write a program to remove an element from a set using remove() and discard(), and explain (in a comment) the difference in behavior when the element does not exist

set1 = {1, 2, 3, 4, 5}
set1.remove(3)  # Removes element 3 from the set
# If element 3 did not exist in the set, this would raise a KeyError
print("Set after remove():", set1)

set1.discard(6)  # Removes element 6 from the set if it exists
# If element 6 did not exist in the set, this would not raise an error
print("Set after discard():", set1)
```

```python
#Write a program to check whether two sets are disjoint (have no elements in common) using isdisjoint().

set1 = {1, 2, 3}
set2 = {4, 5, 6}

is_disjoint = set1.isdisjoint(set2)
print("Are set1 and set2 disjoint?", is_disjoint)
```
